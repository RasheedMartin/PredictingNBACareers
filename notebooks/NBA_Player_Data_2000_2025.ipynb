## Get Data from CSVs and Create Individual DataFrames

```python
from pathlib import Path
import polars as pl
import re


def combine_csvs(dirpath: Path) -> pl.DataFrame | None:
    dirpath.mkdir(parents=True, exist_ok=True)
    
    if dirpath.exists():
        # Load CSVs
        df_list = []
        for file in dirpath.iterdir():
            df = pl.read_csv(file)
            match = re.search(r"(\d{4})_(\d{4})", file.stem)
            
            if match:
                season = f"{match.group(1)}-{match.group(2)[-2:]}"
                df = df.with_columns(
                    pl.lit(season).alias("Season"),
                    pl.col("Player")
                    .str.replace(".", "", literal=True)
                    .str.replace("*", "", literal=True)
                )
                df = df.drop("Player-additional", strict=False)
            df_list.append(df)
    if df_list:
        return pl.concat(df_list, how="diagonal_relaxed")
    return None
```

```python
# Create Advanced Dataframe
dirpath = Path("../data/raw_data/NBA_Player_Data/Advanced_Data")
advanced_df = combine_csvs(dirpath)
advanced_df
```

```python
# Create Per Game Dataframe
dirpath = Path("../data/raw_data/NBA_Player_Data/Per_Game_Data")
per_game_df = combine_csvs(dirpath)
per_game_df
```

```python
# Create Total Player Dataframe
dirpath = Path("../data/raw_data/NBA_Player_Data/Total_Player_Data")
total_player_df = combine_csvs(dirpath)
total_player_df
```

### Lets try to add in the heights and weights and colleges from another dataset

```python
import kagglehub
import shutil 
from pathlib import Path

# Download latest version
path = kagglehub.dataset_download("blitzapurv/nba-players-data-1950-to-2021")

print("Path to dataset files:", path)


dir = Path("../data/raw_data/NBA_Player_Data/")
dir.mkdir(parents=True, exist_ok=True)
shutil.copytree(path, dir, dirs_exist_ok=True)
```

```python
players_df = pl.read_csv(dir / "Players.csv")
```

```python
players_df.head()
```

```python
# Clean up Player names
players_df.columns = list(map(str.title, players_df.columns))
players_df = players_df.with_columns(
    pl.col("Player")
    .str.replace_all(".", "", literal=True)
    .str.replace_all("*", "", literal=True),
    (pl.col("Weight") * 2.20462).round(0).cast(pl.Int64), # Convert to lbs
    pl.col("Collage").alias("Colleges")
).drop("Collage")
```

```python
players_df.head()
```

```python
players_df = players_df.drop("", strict=False)
```

```python
new_master_df = master_df.join(players_df, on="Player", how="left")
```

```python
other_player_df = pl.read_csv(dir / "player_data.csv")
other_player_df = other_player_df[["Player", "Colleges"]]
```

```python
other_player_df = other_player_df.with_columns(
    pl.col("Player")
    .str.replace_all(".", "", literal=True)
    .str.replace_all("*", "", literal=True),
)
```

```python
df = new_master_df.join(other_player_df, on="Player", how="left", coalesce=True)
```

```python
df = df.with_columns(
    pl.coalesce([
        pl.when((pl.col("Colleges").is_null()) | (pl.col("Colleges") == ""))
          .then(None)
          .otherwise(pl.col("Colleges")),
        pl.when((pl.col("Colleges_right").is_null()) | (pl.col("Colleges_right") == ""))
          .then(None)
          .otherwise(pl.col("Colleges_right")),
        pl.lit("Unknown")
    ]).alias("College")
).drop(["Colleges", "Colleges_right"])
```
