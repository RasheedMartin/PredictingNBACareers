### Aggregate the Seasons Stats Data

```python
import polars as pl

df = pl.read_csv("../data/raw_data/NBA_Player_Data/seasons_stats.csv", encoding="utf8-lossy")
df.head()
```

```python
df.max()
```

```python
df.unique()
```

```python
int_cols_to_cast = [
    'MP', '3PA', '3P','ORB',
    'DRB', 'TRB', 'STL',
    'BLK', 'TOV', 'GS']
float_cols_to_cast = [
    'WS/48', 'PER', '3PAr', 'OBPM',
    'DBPM', 'BPM','3P%', 'ORB%',
    'DRB%', 'TRB%', 'AST%', 'STL%',
    'BLK%', 'TOV%', 'USG%', 'VORP'
]

df = df.with_columns(
    [pl.col(col).cast(pl.Int64, strict=False) for col in int_cols_to_cast],
)
df = df.with_columns(
    [pl.col(col).cast(pl.Float64, strict=False) for col in float_cols_to_cast],
)
df
```

```python
df = df.drop_nulls(pl.col("Player"))
df.null_count()
```

```python
df.max()
```

```python
df = df.with_columns(
    pl.col("Player")
    .str.replace_all("*", "", literal=True)
    .str.replace_all(".", "", literal=True)
)
df
```

```python
# Only keep records after 1999
df = df.filter(pl.col("Year") >= 2000)
```

```python
df.max()
```

```python
from utils.utils import nba_teams

df = df.with_columns(
    pl.col("Tm").map_elements(lambda x: nba_teams.get(x, "Unknown"), return_dtype=pl.String).alias("Team_Full_Name")
)
df
```

```python

dirpath = Path("../data/clean_data/NBA_Player_Data")
dirpath.mkdir(parents=True, exist_ok=True)
filepath = Path(dirpath / "nba_list_no_agg.csv")
if dirpath.exists():
    df.write_csv(filepath, separator=",")
```

```python

agg_df = df.group_by("Player").agg(
    pl.all().exclude(pl.String).mean()
)

agg_df = agg_df.drop(["Year", "Age"])
```

```python
filepath = Path(dirpath / "nba_list_agg.csv")
if dirpath.exists():
    agg_df.write_csv(filepath, separator=",")
```
