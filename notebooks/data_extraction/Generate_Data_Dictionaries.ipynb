```python
# Imports
from pathlib import Path
import polars as pl
import json
```

```python
df_gen = Path("../data/clean_data/").rglob("*.csv")
```

```python
from utils.utils import generate_full_polars_data_dictionary

def generate_data_dictionary(file: Path):
    # Load CSV with Polars
    df = pl.read_csv(file, null_values="NA", encoding="utf8-lossy")
    # Generate detailed data dictionary
    data_dict = generate_full_polars_data_dictionary(df)

    # Define output path
    output_path = file.parent / f"{file.stem}_data_dictionary.json"

    # Write to JSON file
    with output_path.open("w", encoding="utf-8") as f:
        json.dump(data_dict, f, indent=4, ensure_ascii=False)

    print(f"✅ Data dictionary written to: {output_path}")
```

```python
for file in df_gen:
    generate_data_dictionary(file)
```

```python
df_gen = Path("../data/raw_data/").rglob("*.csv")
```

```python
for file in df_gen:
    for header in ["Advanced_Data", "Per_Game_Data", "Total_Player_Data", "raw_csvs"]:
        if header in file.parent.parts:
            break
    else:
        generate_data_dictionary(file)
```
